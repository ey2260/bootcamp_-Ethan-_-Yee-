```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler


def transform_data(df: pd.DataFrame, num_cols: list[str] = None) -> pd.DataFrame:
    """Transforms raw features into a model-ready format:

    - Applies Min-Max scaling to numerical columns.
    - One-hot encodes categorical text columns.
    """
    df_transformed = df.copy()

    # 1. Scale numerical features
    if num_cols:
        scaler = MinMaxScaler()
        df_transformed[num_cols] = scaler.fit_transform(df_transformed[num_cols])

    # 2. Encode categorical columns
    cat_cols = df_transformed.select_dtypes(include=["object", "category"]).columns
    df_transformed = pd.get_dummies(
        df_transformed, columns=cat_cols, drop_first=True
    )

    return df_transformed
```
